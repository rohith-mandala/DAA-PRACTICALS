## Linear vs. Binary Search Algorithms

This notebook will implement both linear and binary search algorithms and analyze their time complexity by measuring their execution times on a sample dataset.

```python
def linear_search(arr, target):
    """Implements the linear search algorithm."""
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

print("Linear search function defined.")
```

```python
def binary_search(arr, target):
    """Implements the binary search algorithm (requires a sorted array)."""
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

print("Binary search function defined.")
```

### Test Data Generation

We will generate a large sorted list to test the performance of both algorithms.

```python
import random

# Generate a sorted list of numbers
list_size = 1000000
sorted_list = sorted(random.sample(range(list_size * 2), list_size))

# Choose a target element that exists in the list for a successful search
target_exists = sorted_list[random.randint(0, list_size - 1)]

# Choose a target element that does not exist in the list for an unsuccessful search
target_not_exists = list_size * 2 + 1 # A number guaranteed not to be in the list

print(f"Generated a sorted list of {list_size} elements.")
print(f"Target (exists): {target_exists}")
print(f"Target (does not exist): {target_not_exists}")
```

### Time Complexity Analysis

We will use the `time` module to measure the execution time of both algorithms for successful and unsuccessful searches.

```python
import time

# --- Test Linear Search ---
print("\n--- Linear Search ---")

start_time = time.perf_counter()
linear_result_exists = linear_search(sorted_list, target_exists)
end_time = time.perf_counter()
print(f"Linear Search (target exists): Found at index {linear_result_exists} in {end_time - start_time:.6f} seconds")

start_time = time.perf_counter()
linear_result_not_exists = linear_search(sorted_list, target_not_exists)
end_time = time.perf_counter()
print(f"Linear Search (target not exists): Found at index {linear_result_not_exists} in {end_time - start_time:.6f} seconds")

# --- Test Binary Search ---
print("\n--- Binary Search ---")

start_time = time.perf_counter()
binary_result_exists = binary_search(sorted_list, target_exists)
end_time = time.perf_counter()
print(f"Binary Search (target exists): Found at index {binary_result_exists} in {end_time - start_time:.6f} seconds")

start_time = time.perf_counter()
binary_result_not_exists = binary_search(sorted_list, target_not_exists)
end_time = time.perf_counter()
print(f"Binary Search (target not exists): Found at index {binary_result_not_exists} in {end_time - start_time:.6f} seconds")
```

### Conclusion

As observed from the execution times, binary search is significantly faster than linear search for large datasets because of its logarithmic time complexity ($O(\log n)$) compared to linear search's linear time complexity ($O(n)$). This efficiency comes at the cost of requiring the input array to be sorted.
