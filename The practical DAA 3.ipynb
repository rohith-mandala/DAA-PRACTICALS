## Max Heap Sort Algorithm Implementation

Heap sort is a comparison-based sorting technique based on binary heap data structure. It is similar to selection sort where we first find the maximum element and place it at the end. We repeat the same process for the remaining elements.

Here, we will implement the Max Heap Sort algorithm step-by-step.

### 1. `max_heapify` Function

This function maintains the max-heap property. If the root node is smaller than its children, it swaps the root with the largest child and recursively calls `max_heapify` on the affected subtree.

```python
def max_heapify(arr, n, i):
    """Maintains the max-heap property for a subtree rooted at index i."""
    largest = i  # Initialize largest as root
    left = 2 * i + 1  # left child
    right = 2 * i + 2  # right child

    # See if left child of root exists and is greater than root
    if left < n and arr[largest] < arr[left]:
        largest = left

    # See if right child of root exists and is greater than current largest
    if right < n and arr[largest] < arr[right]:
        largest = right

    # Change root, if needed
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # swap

        # Heapify the root.
        max_heapify(arr, n, largest)

print("`max_heapify` function defined.")
```

### 2. `build_max_heap` Function

This function converts an arbitrary array into a max heap. It does this by calling `max_heapify` on all non-leaf nodes, starting from the last non-leaf node and moving up to the root.

```python
def build_max_heap(arr):
    """Builds a max heap from an unsorted array."""
    n = len(arr)
    # Start from the last non-leaf node and heapify upwards
    for i in range(n // 2 - 1, -1, -1):
        max_heapify(arr, n, i)

print("`build_max_heap` function defined.")
```

### 3. `heap_sort` Function

This is the main heap sort function. It first builds a max heap from the input array. Then, it repeatedly extracts the maximum element (root of the heap) and places it at the end of the array, then re-heapifies the remaining elements.

```python
def heap_sort(arr):
    """Sorts an array using the Max Heap Sort algorithm."""
    n = len(arr)

    # Build a max heap
    build_max_heap(arr)

    # One by one extract elements
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]  # Swap root (max element) with last element
        max_heapify(arr, i, 0)  # Call max_heapify on the reduced heap

print("`heap_sort` function defined.")
```

### 4. Demonstration of Max Heap Sort

Let's test our `heap_sort` function with an example array.

```python
import random

# Example usage:
unsorted_array = [random.randint(1, 100) for _ in range(15)]
print("Original array:", unsorted_array)

heap_sort(unsorted_array)

print("Sorted array:", unsorted_array)
```

### Time and Space Complexity Analysis

*   **Time Complexity:** The time complexity of Heap Sort is $O(n \log n)$ in all cases (worst, average, and best). This is because `build_max_heap` takes $O(n)$ time, and the loop for extracting elements runs `n-1` times, with each `max_heapify` operation taking $O(\log n)$ time.
*   **Space Complexity:** The space complexity of Heap Sort is $O(1)$ because it sorts in-place, meaning it doesn't require any additional memory proportional to the input size (excluding the input array itself).
