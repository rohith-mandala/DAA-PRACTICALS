```python
import time

def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
my_list = [64, 34, 25, 12, 22, 11, 90]

start_time = time.time()
sorted_list = bubble_sort(my_list.copy())
end_time = time.time()

print("Original array:", my_list)
print("Sorted array:", sorted_list)
print(f"Execution time: {end_time - start_time:.6f} seconds")
```

```python
import time

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
my_list_selection = [64, 34, 25, 12, 22, 11, 90]

start_time_selection = time.time()
sorted_list_selection = selection_sort(my_list_selection.copy())
end_time_selection = time.time()

print("Original array (Selection Sort):", my_list_selection)
print("Sorted array (Selection Sort):", sorted_list_selection)
print(f"Execution time (Selection Sort): {end_time_selection - start_time_selection:.6f} seconds")
```

```python
import time

def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2 
        L = arr[:mid]        
        R = arr[mid:]      

        merge_sort(L)        
        merge_sort(R)        

        i = j = k = 0
        while i < len(L) and j < len(R):
            if L[i] < R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1
        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1

        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1
    return arr
    
my_list_merge = [64, 34, 25, 12, 22, 11, 90, 5, 88]

start_time_merge = time.time()
sorted_list_merge = merge_sort(my_list_merge.copy()) 
end_time_merge = time.time()

print("Original array (Merge Sort):", my_list_merge)
print("Sorted array (Merge Sort):", sorted_list_merge)
print(f"Execution time (Merge Sort): {end_time_merge - start_time_merge:.6f} seconds")
```

```python
import time

def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    else:
        pivot = arr[len(arr) // 2]
        left = [x for x in arr if x < pivot]
        middle = [x for x in arr if x == pivot]
        right = [x for x in arr if x > pivot]
        return quick_sort(left) + middle + quick_sort(right)
        
my_list_quick = [64, 34, 25, 12, 22, 11, 90, 5, 88, 77, 1]

start_time_quick = time.time()
sorted_list_quick = quick_sort(my_list_quick.copy()) 
end_time_quick = time.time()

print("Original array (Quick Sort):", my_list_quick)
print("Sorted array (Quick Sort):", sorted_list_quick)
print(f"Execution time (Quick Sort): {end_time_quick - start_time_quick:.6f} seconds")
```
