```python
from collections import deque
from time import perf_counter_ns


class Graph:
    def __init__(self, vertices):
        self.V = vertices
        self.adj = [[] for _ in range(vertices)]

    def add_edge(self, u, v):
        self.adj[u].append(v)
        self.adj[v].append(u)  # Remove this line for a directed graph

    def _dfs_util(self, vertex, visited, traversal):
        visited[vertex] = True
        traversal.append(vertex)

        for neighbor in self.adj[vertex]:
            if not visited[neighbor]:
                self._dfs_util(neighbor, visited, traversal)

    def dfs(self, start):
        visited = [False] * self.V
        traversal = []
        self._dfs_util(start, visited, traversal)
        return traversal

    def bfs(self, start):
        visited = [False] * self.V
        traversal = []
        queue = deque([start])
        visited[start] = True

        while queue:
            node = queue.popleft()
            traversal.append(node)

            for neighbor in self.adj[node]:
                if not visited[neighbor]:
                    visited[neighbor] = True
                    queue.append(neighbor)

        return traversal


def main():
    vertices = int(input("Enter number of vertices: "))
    graph = Graph(vertices)

    edges = int(input("Enter number of edges: "))
    print("Enter edges (u v):")

    for _ in range(edges):
        u, v = map(int, input().split())
        graph.add_edge(u, v)

    start = int(input("Enter starting vertex: "))

    start_dfs = perf_counter_ns()
    dfs_traversal = graph.dfs(start)
    dfs_time = perf_counter_ns() - start_dfs
    print("\nDFS Traversal:", *dfs_traversal)

    start_bfs = perf_counter_ns()
    bfs_traversal = graph.bfs(start)
    bfs_time = perf_counter_ns() - start_bfs
    print("\nBFS Traversal:", *bfs_traversal)

    print("\nExecution Time:")
    print(f"DFS: {dfs_time} ns")
    print(f"BFS: {bfs_time} ns")


if __name__ == "__main__":
    main()
```
